```python
# ==============================================================================
# НАУЧНО-ЭКСПЕРИМЕНТАЛЬНЫЙ СТЕНД: СРАВНЕНИЕ METOДОВ PEOPLE COUNTING
# СТРОГОЕ СООТВЕТСТВИЕ МЕТОДОЛОГИИ МЕЖДУНАРОДНЫХ СТАНДАРТОВ (РИНЦ / SCOPUS)
# ==============================================================================

# 1. Установка и импорт верифицированных библиотек
!pip install -q ultralytics scipy opencv-python pandas matplotlib

import os
import re
import time
import platform
import numpy as np
import pandas as pd
import scipy.io
import cv2
import torch
import matplotlib.pyplot as plt
from ultralytics import YOLO
from pathlib import Path

print("📦 Шаг 1: Инициализация окружения и проверка путей датасета...")

# Автоматическое определение пути к загруженному кэшу KaggleHub
import kagglehub
try:
    DATASET_PATH = Path(kagglehub.dataset_download("chaozhuang/mall-dataset"))
except Exception:
    # Резервный статический путь, если kagglehub вызван повторно
    DATASET_PATH = Path("/root/.cache/kagglehub/datasets/chaozhuang/mall-dataset/versions/3")

GT_PATH = DATASET_PATH / "mall_gt.mat"

if not GT_PATH.is_file():
    raise FileNotFoundError(f"❌ Критическая ошибка: Файл разметки не найден по пути: {GT_PATH}")





```

```python
# 2. Рекурсивный поиск и строгая естественная (числовая) сортировка кадров
all_jpg_paths = list(DATASET_PATH.rglob("*.jpg")) + list(DATASET_PATH.rglob("*.JPG"))
all_jpg_paths = list(set(all_jpg_paths)) # Удаление дубликатов путей

def natural_sort_key(path_obj):
    """Обеспечивает строго хронологический порядок файлов (seq_2 перед seq_10)"""
    return [int(text) if text.isdigit() else text.lower() for text in re.split(r'(\d+)', str(path_obj.name))]

all_jpg_paths = sorted(all_jpg_paths, key=natural_sort_key)

# Загрузка оригинальной матрицы MATLAB Ground Truth
mat = scipy.io.loadmat(str(GT_PATH))
all_gt_counts = np.asarray(mat["count"]).reshape(-1)

print(f"✔️ Найдено физических кадров изображений: {len(all_jpg_paths)}")
print(f"✔️ Найдено эталонных аннотаций в векторе: {len(all_gt_counts)}")

if len(all_jpg_paths) != len(all_gt_counts):
    raise ValueError("❌ Аномалия данных: Несоответствие количества кадров и эталонных меток разметки!")
```

```python
# 3. Фиксация академического тестового протокола (Кадры 801 - 2000, N = 1200)
TEST_START = 800
TEST_END = 2000

test_paths = all_jpg_paths[TEST_START:TEST_END]
true_counts = all_gt_counts[TEST_START:TEST_END].astype(np.int64)
num_test_frames = len(test_paths)

print(f"✔️ Сформирована репрезентативная тестовая выборка. Объем N = {num_test_frames} кадров.")
assert len(all_jpg_paths) == len(all_gt_counts), (
    "Количество изображений не совпадает с количеством аннотаций"
)

assert len(test_paths) == 1200, (
    f"Ожидалось 1200 кадров, найдено {len(test_paths)}"
)

assert len(true_counts) == len(test_paths), (
    "Количество эталонных значений не совпадает с количеством кадров"
)

print("Первые 10 файлов:")
for p in test_paths[:10]:
    print(p.name)

print("Последние 10 файлов:")
for p in test_paths[-10:]:
    print(p.name)
```

```python
from scipy.io import loadmat

mat = loadmat(str(GT_PATH))

print("Переменные в mall_gt.mat:")
print(mat.keys())

for key in mat:
    if not key.startswith("__"):
        value = mat[key]
        print(
            f"{key}: тип={type(value)}, "
            f"shape={getattr(value, 'shape', None)}, "
            f"dtype={getattr(value, 'dtype', None)}"
        )
```

```python
import numpy as np
from scipy.io import loadmat

mat = loadmat(str(GT_PATH))

frames = mat["frame"].flatten()
counts = mat["count"].flatten()

print("Количество записей frame:", len(frames))
print("Количество значений count:", len(counts))

print("\nПервые 5 записей frame:")
for i in range(5):
    print(f"Индекс {i}: тип={type(frames[i])}, значение={frames[i]}")

print("\nПервые 10 значений count:")
print(counts[:10])
```

```python
# 4. Предварительное кэширование изображений в RAM для исключения I/O оверхеда из FPS
print("\n💾 Шаг 2: Кэширование тестовых кадров в оперативную память...")
images_ram = []
for p in test_paths:
    img = cv2.imread(str(p))
    if img is None:
        raise ValueError(f"Не удалось прочитать файл кадра: {p}")
    images_ram.append(img)
print(f"✔️ Успешно загружено в RAM: {len(images_ram)} матриц. Разрешение кадра: {images_ram[0].shape}")
```

```python
# 5. Определение аппаратных средств и инициализация весов YOLOv8x
DEVICE = 0 if torch.cuda.is_available() else "cpu"
print(f"\n🧠 Шаг 3: Инициализация YOLOv8x на вычислительном устройстве: {DEVICE}")
model = YOLO("yolov8x.pt")

CONF_THRESHOLD = 0.25  # Стандартизированный зафиксированный порог уверенности
IMG_SIZE_INFERENCE = 640

# Предварительный холостой прогрев (Warm-up) графического ускорителя GPU
for _ in range(10):
    _ = model.predict(source=images_ram[0], classes=[0], conf=CONF_THRESHOLD, imgsz=IMG_SIZE_INFERENCE, device=DEVICE, verbose=False)
if torch.cuda.is_available():
    torch.cuda.synchronize()
```

```python
# ==============================================================================
# ЭТАП A: Честный сквозной инференс и бенчмарк YOLOv8x (на GPU/DEVICE)
# ==============================================================================
print("⏳ Запуск сквозного тестирования YOLOv8x...")
yolo_counts = []

if torch.cuda.is_available():
    torch.cuda.synchronize()
t_start_yolo = time.perf_counter()

for img in images_ram:
    result = model.predict(source=img, classes=[0], conf=CONF_THRESHOLD, imgsz=IMG_SIZE_INFERENCE, device=DEVICE, verbose=False)[0]
    # Извлекаем количество предсказанных рамок для класса 'person'
    yolo_counts.append(0 if result.boxes is None else len(result.boxes))

if torch.cuda.is_available():
    torch.cuda.synchronize()
t_elapsed_yolo = time.perf_counter() - t_start_yolo
fps_yolo = num_test_frames / t_elapsed_yolo
```

```python
# ==============================================================================
# ЭТАП Б: Честный сквозной инференс и бенчмарк полного OpenCV пайплайна (на CPU)
# ==============================================================================
print("⏳ Запуск сквозного тестирования Canny-based Contour Baseline...")
opencv_counts = []

def run_full_opencv_pipeline(img):
    """Полный вычислительный конвейер OpenCV, эквивалентный задаче подсчета"""
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    edges = cv2.Canny(blurred, 30, 100)
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))
    dilated = cv2.dilate(edges, kernel, iterations=1)
    contours, _ = cv2.findContours(dilated, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # Подсчет геометрических контуров с площадью выше шумового порога
    valid_blobs = sum(1 for cnt in contours if cv2.contourArea(cnt) > 30)
    return valid_blobs

t_start_cv = time.perf_counter()

for img in images_ram:
    opencv_counts.append(run_full_opencv_pipeline(img))

t_elapsed_cv = time.perf_counter() - t_start_cv
fps_opencv = num_test_frames / t_elapsed_cv

# Преобразование результатов в массивы NumPy
yolo_counts = np.array(yolo_counts, dtype=np.int64)
opencv_counts = np.array(opencv_counts, dtype=np.int64)
```

```python
# ==============================================================================
# ЭТАП В: Расчет расширенных статистических метрик качества (MAE, RMSE, Bias)
# ==============================================================================
print("\n📊 Шаг 4: Расчет математических метрик эффективности алгоритмов...")

def compute_academic_metrics(y_true, y_pred):
    errors = y_pred - y_true
    mae = np.mean(np.abs(errors))
    rmse = np.sqrt(np.mean(errors ** 2))
    bias = np.mean(errors) # Отрицательный Bias означает недоучет, положительный - переучет
    return {"MAE": mae, "RMSE": rmse, "Bias": bias}

metrics_yolo = compute_academic_metrics(true_counts, yolo_counts)
metrics_opencv = compute_academic_metrics(true_counts, opencv_counts)
```

```python
# ==============================================================================
# ЭТАП Г: Экспорт результатов в CSV-архивы и вывод сводной таблицы
# ==============================================================================
print("\n💾 Шаг 5: Сохранение результатов и экспорт структурированных таблиц...")

# 1. Покадровый детальный лог ошибок
df_frames = pd.DataFrame({
    "frame_index": np.arange(TEST_START + 1, TEST_END + 1),
    "ground_truth": true_counts,
    "yolo_predict": yolo_counts,
    "opencv_predict": opencv_counts,
    "yolo_abs_error": np.abs(yolo_counts - true_counts),
    "opencv_abs_error": np.abs(opencv_counts - true_counts)
})
df_frames.to_csv("mall_counting_detailed_results.csv", index=False)

# 2. Итоговая таблица для публикации в LaTeX
df_summary = pd.DataFrame([
    {"Метод": "Canny-based Contour Baseline", "MAE": metrics_opencv["MAE"], "RMSE": metrics_opencv["RMSE"], "Bias": metrics_opencv["Bias"], "FPS": fps_opencv, "Устройство": "CPU"},
    {"Метод": "YOLOv8x (Pre-trained COCO)", "MAE": metrics_yolo["MAE"], "RMSE": metrics_yolo["RMSE"], "Bias": metrics_yolo["Bias"], "FPS": fps_yolo, "Устройство": str(DEVICE)}
])
df_summary.to_csv("mall_counting_summary.csv", index=False)

print("\n📈 СВОДНАЯ ТАБЛИЦА РЕЗУЛЬТАТОВ ЭКСПЕРИМЕНТА:")
print(df_summary.to_string(index=False))
```

```python
# ==============================================================================
# ЭТАП Д: Генерация и сохранение академического графика распределения ошибок
# ==============================================================================
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# Верхний график: Динамика изменения истинных и предсказанных значений
ax1.plot(df_frames["frame_index"], df_frames["ground_truth"], label="Ground Truth (Эталон)", color="#2c3e50", linewidth=2)
ax1.plot(df_frames["frame_index"], df_frames["yolo_predict"], label=f"YOLOv8x (MAE: {metrics_yolo['MAE']:.2f})", color="#2ecc71", alpha=0.8)
ax1.plot(df_frames["frame_index"], df_frames["opencv_predict"], label=f"OpenCV Baseline (MAE: {metrics_opencv['MAE']:.2f})", color="#e74c3c", alpha=0.6)
ax1.set_ylabel("Количество людей (чел.)", fontsize=11)
ax1.set_title("Сравнительный анализ временных последовательностей подсчета объектов", fontsize=13, fontweight='bold')
ax1.legend(loc="upper right", frameon=True)

# Нижний график: Модуль абсолютной ошибки по ходу видеопотока
ax2.fill_between(df_frames["frame_index"], df_frames["opencv_abs_error"], color="#e74c3c", alpha=0.2, label="Абсолютная ошибка OpenCV")
ax2.fill_between(df_frames["frame_index"], df_frames["yolo_abs_error"], color="#2ecc71", alpha=0.2, label="Абсолютная ошибка YOLOv8x")
ax2.set_xlabel("Индекс кадра в видеопоследовательности", fontsize=11)
ax2.set_ylabel("Абсолютная ошибка (|Y - Ŷ|)", fontsize=11)
ax2.legend(loc="upper right", frameon=True)

plt.tight_layout()
plt.savefig("metrics_comparison.png", dpi=300)
plt.show()

print(f"\n📂 Успешно сгенерированы файлы для Overleaf:\n 1. 'metrics_comparison.png' (График высокого разрешения)\n 2. 'mall_counting_summary.csv' (Сводная таблица метрик)")
```
